## Class with constructor function

```python
The __init__() method is used to assign values to object properties.
Values are passed as arguments to the constructor, and self is used to store those values as instance variables inside the current object.
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name,
        self.marks = marks
        
```

```python
s1 = Student("Karan", 97)
```

```python
s2 = Student("Arjun", 99)
```

```python
print(s1.name)
print(s1.marks)
```

## Self


```python
Self refers to the current object of a class.

It allows an object to access its own variables and methods.

Think of it this way:
- Class = Blueprint
- Object = Actual item created from the blueprint.
- self = Refers to the particular object currently being used.

self.name = name	Assigns a value to an instance variable
```

## Excercise 1:

```python
## Eg: Create student class that takes name and marks of 3 subjects as arguments in constructor. Then create a method to print the average.
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def get_avg(self):
        sum = 0
        for val in self.marks:
            sum += val
        print("Hi,", self.name, "Your avg score is:", sum/3)
        
```

```python
s1 = Student("Max",[99,98,97])
s1.get_avg()
```

## Excercise 4 :

Exercise 4: BankAccount – Methods inside a class
Difficulty: Beginner+
Create a class called BankAccount.

Requirements:
1. Create a constructor (__init__) accepting account_holder and balance.
2. Store both using self.
3. Create a method called deposit(amount):
   - Add the amount to the existing balance.
4. Create a method called display_balance():
   - Print the account holder and current balance.
5. Create an object and call both methods.

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def display_balance(self):
        print("Account_holder:", self.account_holder, self.balance)
            
```

```python
cust1 = BankAccount("Max", 4000)
```

```python
cust1.display_balance()
```

```python
cust1.deposit(2000)
```

```python
print("After deposit:")
cust1.display_balance()
```
